```python
def insertionSort(ar):

    for i in range(1,len(ar)):

        element = ar[i]
        j = i - 1
        while j >= 0 and ar[j] > element:
            ar[j+1] = ar[j]
            j -= 1
        ar[j + 1] = element
    return ar

print(insertionSort( [13, 46, 24, 52, 20, 9]))
```

```python
def findlong(arr):
    maxlen = 0

    for i in range(len(arr)):
        cursum = 0
        for j in range(i,len(arr)):
            cursum += arr[j]
            if cursum == 0:
                maxlen = max(j - i+ 1,maxlen)
    return maxlen

    

print(findlong([6, -2, 2, -8, 1, 7, 4, -10]))
        
```
