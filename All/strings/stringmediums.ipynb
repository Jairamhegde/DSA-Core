```python
def repeatedStringMatch(a, b):
        def check(a,b):
            mod = 10**9+7
            currentSum = 0
            hashcode = 0
            d = {
                "a":1,
                "b":2,
                "c":3,
                "d":4,
                "e":5,
                "f":6,
                "g":7,
                "h":8,
                "i":9,
                "j":10,
                "k":11,
                "l":12,
                "m":13,
                "n":14,
                "o":15,
                "p":16,
                "q":17,"r":18,"s":19,"t":20,"u":21,"v":22,"w":23,"x":24,"y":25,"z":26
            }
            base = 26
            highiestpower =pow(base,len(a) ,mod)
            
            for i in range(len(a)):
                val = d[a[i]]
                hashcode = (hashcode * base + val)%mod

            j = 0
            k = 0
            while j < len(b):
                val = d[b[j]]
                currentSum = (currentSum * base + val)%mod

                if j - k+1 > len(a):
                    val = d[b[k]]
                    currentSum =( (currentSum - val * highiestpower)%mod)
                    k += 1
                
                if j - k + 1 == len(a):
                    if currentSum == hashcode:
                        return True
                j += 1
            
            return False
        count = 0
        n = len(a)
        repeated = []
        while len("".join(repeated)) < len(b):
            repeated .append(a)
            count += 1

        if check(b,"".join(repeated)):
            return count
        repeated.append(a)
        count += 1
        if check(b,"".join(repeated)):
            return count
        return -1

print(repeatedStringMatch("abcd","ababcdcd"))
```

*Swap the vowels of the string

```python
def reverseVowels(s):
        vowels = {"a","e","i","o","u","A","E","I","O","U"}
        new = list(s)
        print(new)
        n = len(new)
        i = 0
        j = len(new)-1
        while i < j:
            while i < j and new[i] not in vowels:
                i += 1
            while j > i and new[j] not in vowels:
                j -= 1
            new[i],new[j] = new[j],new[i]
            i += 1
            j -= 1
        return "".join(new)
print(reverseVowels("leetcode"))
```

```python
* Zigzag conversion fo the string
Input: s = "PAYPALISHIRING", numRows = 3
Output: "PAHNAPLSIIGYIR"
P   A   H   N
A P L S I I G
Y   I   R
```

```python
def zigzag(s,numRows):

    if numRows == 1 | len(s) == 1:
        return s
    new = [""] * numRows
    
    cycle = 2 *numRows - 2

    for i, ch in enumerate(s):
        pos = i % cycle
        row = pos if pos < numRows else cycle - pos
        new[row] += ch

    return "".join(new)
print(zigzag("PAYPALISHIRING",3))
    
```

* 443. String Compression

```python
def compress( chars):
        i = 0
        n = len(chars)
        j = 0
        while j < n:
                char = chars[j]
                count = 0

                while j < n and chars[j] == char:
                        j += 1
                        count += 1
                chars[i] = char
                i += 1
                if count > 1:
                        chars[i] = count
                        i += 1
        return i   

print(compress(["a","a","b","b","c","c","c"]))
        

        
```

```python
def thrreesum(array):
    ar = sorted(array)
    print(ar)
    new = []
    def helper(index,current,cursum):
        if len(current) == 3:
            if cursum == 0:
                new.append(current[:])
            return
        for i in range(index,len(ar)):
            if i > index and ar[i] == ar[i-1]:  # skip duplicate
                continue
            if cursum > 0:
                return
            current.append(ar[i])
            cursum += ar[i]
            helper(i+1,current,cursum)
            item = current.pop()
            cursum -= item
    helper(0,[],0)
    return new

print(thrreesum([-1,0,1,2,-1,-4]))


    


    
```

```python
def three_sum(nums):
    new = sorted(nums)
    
    n = len(new)
    resulting_set = set()
    for i in range(n):
        stored_set = set()
        for j in range(i+1,n):
            required = 0 - (new[i] + new[j])
            if required in stored_set:
                triplet = tuple(sorted((new[i],new[j],required)))
                resulting_set.add(triplet)   
            stored_set.add(new[j])
    return list(resulting_set)

print(three_sum([-1,0,1,2,-1,-4]))


    
```

```python
def longestPalindrome(s):
    if not s:
        return 0
    map = {}
    addfre = False
    total = 0
    for i in range(len(s)):
        map[s[i]] = map.get(s[i],0)+1

    for item in map.values():
        if item % 2 == 0:
            total += item
        else:
            total += item - 1
            addfre = True
    
    return total+1 if addfre == True else total 


print(longestPalindrome("abccccdd"))
```

```python
def reverseWords(s):
    new = list(s)
    j = 0
    n = len(new)
    for i in range(n):
        if new[i] == " " or  i == n-1:
            l = j
            if i == n-1:
                r = i
            else:
                r = i-1
            while l < r:
                new[l],new[r] = new[r],new[l]
                l += 1
                r -= 1
            j = i + 1
    return "".join(new)


print(reverseWords(""))
    
```

```python
def allCellsDistOrder(rows, cols, rCenter, cCenter):
    new = []
    for i in range(rows):
        for j in range(cols):
            current =[i,j]
            new.append(current)
    new = sorted(new,key = lambda x :abs(rCenter - x[0]) + abs(cCenter-x[1]))
    return new
print(allCellsDistOrder(3,4,0,1))
        
```

```python
def merge_two_sorted(a1,a2):
    left = 0
    mid = len(a1) 
    right = len(a1)
    a1.extend(a2)
    while left < mid and right < len(a1):
        if a1[left] > a1[right]:
            a1[left],a1[right] = a1[right],a1[left]
            left += 1
        else:
            right += 1
    return a1


print(merge_two_sorted([1,2,3,0,0,0],[2,5,6]))



```
