BINARY SEARCH

```python
def binarySearch(n,k):
    mid = len(n)// 2
    i = 0
    j = len(n)-1
    while(i < j):
        if n[mid] == k:
            return mid
        elif k > n[mid]:
            i = mid
            mid = (i +j )//2
        else:
            mid = j
            mid = (i + j) // 2
    return -1
print(binarySearch([1,2,3,4,5,6,7,8,9],7))
```

BINARY SEARCH USING RECURSION

```python
def binarySearch(n,low,high,k):
    if low >= high:
        return -1
    mid = (low + high)// 2
  
    if n[mid] == k:
        return mid
    elif k > n[mid]:
       return binarySearch(n,mid,high,k)
    else:
       return binarySearch(n,low,mid,k)
ary = [1,2,3,4,5,6,7,8,9]
print(binarySearch(ary,0,len(ary) - 1,7))
```

FIND THE LOWER BOUND

```python
def lowerBound(n,x):
    i = 0
    j = len(n) - 1
    lowerBound = len(n)
    while(i <= j):
        mid = (i + j)//2
        if n[mid] >= x:
            lowerBound = min(lowerBound,mid)
            j = mid - 1
        else:
            i = mid + 1
    return lowerBound
print(lowerBound([1,2,2,3],2))
        
```

```python
def upperBound(n,x):
    i = 0
    j = len(n) - 1
    upperBound = 0
    while(i <= j):
        mid = (i + j)//2
        if n[mid] > x:
            upperBound = max(upperBound,n[mid])
            j = mid - 1
        else:
            i = mid + 1
    return upperBound
print(upperBound([3,5,8,15,19],15))
```

SEARCH FOR THE FIRST AND LAST OCCURENCE OF THE TARGET ELEMENT
Input: nums = [5,7,7,8,8,10], target = 8

```python
def searchFor(n,k):
    lastOc = -1
    for i in range(len(n)):
        if n[i] == k:
            lastOc = i
    return lastOc
print(searchFor([3, 4, 13, 13, 13, 20, 40],13))
```

search for the element in a rotated array , and achieve O(log n) time complexity

```python
def search( nums, target):
        # Set initial search space
        low = 0
        high = len(nums) - 1

        # Run loop while valid search space exists
        while low <= high:
            # Find the middle index
            mid = (low + high) // 2

            # If target found at mid, return index
            if nums[mid] == target:
                return mid
        

            # Check if left half is sorted
            if nums[low] <= nums[mid]:
                # If target lies in left half
                if nums[low] <= target < nums[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            else:
                # Right half is sorted
                if nums[mid] < target <= nums[high]:
                    low = mid + 1
                else:
                    high = mid - 1

        # Target not found
        return -1

print(search([1,0,1,1,1],0))
```

Given a sorted array nums and an integer x. Find the floor and ceil of x in nums. The floor of x is the largest element in the array which is smaller than or equal to x. The ceiling of x is the smallest element in the array greater than or equal to x. If no floor or ceil exists, output -1.


Example 1

Input : nums =[3, 4, 4, 7, 8, 10], x= 5

Output: 4 7

Explanation: The floor of 5 in the array is 4, and the ceiling of 5 in the array is 7.

```python
def cielFloar(nums,x):
        low = 0
        high = len(nums) - 1
        ciel = float('inf')
        floar = 0
        while (low <= high):
            mid = (low + high)//2
            if nums[mid] == x:
                return nums[mid],nums[mid]
            elif nums[mid] > x:
                ciel = nums[mid]
                high = mid -1
            else:
                floar = nums[mid]
                low = mid + 1
        return floar,ciel


print(cielFloar([3, 4, 4, 7, 8, 10],9))
    
```

FIND MINIMUM ELEMENT FROM A ROTATED ARRAY

```python
def minElement(nums):
    low = 0
    high = len(nums)-1
    minElement = float('inf')
    while(low <= high):
        mid = (low + high) // 2
        if nums[low] < nums[high]:
            minElement = min(minElement,nums[low])
            break

        if nums[low] <= nums[mid]:
            minElement = min(minElement,nums[low])
            low = mid + 1
        else: 
            minElement = min(minElement,nums[mid])  
            high = mid - 1
    return minElement
            

minElement([4,5,6,7,0,1,2])
```

FIND THE UNIQUE ELEMENT IN THE ARRAY WHICH APPEARS AXACTLY ONE TIME WERE OTHER ELEMENT OCCURS 2 TIMES

```python
def findUnique(nums):
        low = 0
        high= len(nums) - 1
        if len(nums) <=1:
            return nums[0]
        if nums[0] != nums[1]:
            return nums[0]
        if nums[len(nums)-1] != nums[len(nums)-2]:
            return nums[len(nums)-1]
        while(low <= high):
            mid = (low + high)//2
            
            if nums[mid] != nums[mid-1] and nums[mid] != nums[mid+1]:
                return nums[mid]
            elif nums[mid] == nums[mid-1]:
                if mid % 2 != 0:
                    low = mid + 1
                else:
                    high = mid -1
            elif nums[mid] == nums[mid+1]:
                if mid % 2 == 0:
                    low = mid + 1
                else:
                    high = mid - 1
        return -1
            
print(findUnique([2, 2 , 3 , 3 , 3 , 3 , 4, 4,5]))
```
