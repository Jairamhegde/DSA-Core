```python


# Recursive function to count good numbers
def count_good_numbers(index, n):
    MOD = 10**9 + 7
    # Base case: if we've reached the end of the string
    if index == n:
        return 1

    result = 0
    # Even index: Use even digits
    if index % 2 == 0:  
        even_digits = [0, 2, 4, 6, 8]
        for digit in even_digits:
            result = (result + count_good_numbers(index + 1, n)) % MOD
    # Odd index: Use prime digits        
    else:
        prime_digits = [2, 3, 5, 7]
        for digit in prime_digits:
            result = (result + count_good_numbers(index + 1, n)) % MOD
    return result

print(count_good_numbers(0,50))
```

```python
def findPow(n, p):
    mod = 10**9 + 7
    n = n % mod
    
    if p == 0: return 1
    if p == 1: return n
    
    # 1. Prepare variables for next step
    next_p = p // 2
    next_n = (n * n) % mod
    
    # 2. Check odd/even right at the return statement
    if p % 2 != 0:
        # Odd case: Multiply by n
        return (n * findPow(next_n, next_p)) % mod
    else:
        # Even case: Just return the result directly (No multiplying by 1!)
        return findPow(next_n, next_p) % mod
    
print(findPow( 2.0000,10))
```

Function to calculate the power of the given number

```python
def power(n,p):
    mod =10**9 + 7
    def helper(n,count,sum,mod):    
        if count == p:
            return sum
        sum = sum * n % mod
        return helper(n,count+1,sum,mod)
    
    return helper(n,0,1,mod)
print(power(1000,1000))
```

```python
def geneate(nums,target):
    new = []
    current = []
    def helper(new,current,index,currentSum):
        if index == len(nums):
            if currentSum == target:
                new.append(current[:])
            return
        currentSum += nums[index]
        current.append(nums[index])
        helper(new,current,index+1,currentSum)  
        item = current.pop()
        currentSum -= item

        helper(new,current,index+1,currentSum)
        return new

    helper(new,current,0,0)
    return new

print(geneate([10,1,2,7,6,1,5],8))
```

```python
def subsetSum(nums):

    def helper(sumarr,current,index,currentsum):
        sumarr.append(currentsum)
        for i in range(index,len(nums)):
            if i > index and nums[i] == nums[i-1]:
                continue

            current.append(nums[i])
            currentsum += nums[i]
            helper(sumarr,current,i+1,currentsum)
            item = current.pop()
            currentsum -= item
    sumarr = []
    helper(sumarr,[],0,0)
    return sorted(sumarr)
print(subsetSum([5,2,1]))
```

```python
def generate(k,n):
    new = []
    def helper(current):
        if len(current) == k:
            new.append(current[:])
            return
        for i in range(1,k+1):
            if i in current:
                continue
            current.append(i)
            helper(current)
            current.pop()
    helper([])
    return new[n]

print(generate(3,3))
```

* Find Kth permutation of the give len k elements, without manually calculating all the permutations

```python
import math
def findkthpermutation(n,k):
    fact = 1
    elements = []
    for i in range(1,n):
        elements.append(i)
        fact *= i
    elements.append(n)
    res = []
    k = k - 1
    while elements:
        index = k // fact

        res.append(elements[index])
        elements.remove(elements[index])

        if not elements:
            break
        k %= fact
        fact //= len(elements)
    return res


print(findkthpermutation(4,9))
```
