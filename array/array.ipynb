FINDING MEJORITY ELEMENTS IN AN ARRAY WHICH HAS THE COUNT GREATER THAN N/2


```python
# -------------------------BRUTEFORCE
def mejority(n):
    for i in range(len(n)):
        count = 0
        for j in range(i,len(n)):
            if n[j] == n[i]:
                count += 1
        if count > (len(n)//2):
            return n[j]

print(mejority([7, 0, 0, 1, 7, 7, 2, 7, 7]))

# --------------------------OPTIMAL
def mejority(nums):
        map = {}
        for i in nums:
            map[i] = map.get(i,0) + 1
        for index,value in map.items():
            if value > (len(nums) // 2):
                return index
```

FIND THE MAXIMUM POSSIBLE SUM OF THE SUB ARRAY

```python
def maxSum(n):
    max_sum =float('-inf')
    for i in range(len(n)):
        sum = 0
        for j in range(i,len(n)):
            sum += n[j]
            if sum > max_sum:
                max_sum = sum
    return max_sum
print(maxSum([20,30,-1,50,10 ]))


# -----------------------------OPTIMAL

def maxSumm(n):
    sum = 0
    max_sum =0

    for i in range(len(n)):
        sum += n[i]
        if sum > max_sum:
            max_sum = sum

        if sum < 0: #is the sum becomes -ve than reset the sum to zero that is the main key
            sum = 0
    return max_sum
print(maxSumm([1,1,1,-10,-1,20,30 ]))
```

There’s an array ‘A’ of size ‘N’ with an equal number of positive and negative elements. Without altering the relative order of positive and negative elements, you must return an array of alternately positive and negative values.
arr[] = {1,2,-3,-1,-2,-3}, N = 6
Output:
1 -3 2 -1 3 -2

```python

def organizeElements(n):
    ar1 =[0]*len(n)
    possitive_idx = 0
    negetive_idx = 1
    
    for j in range(len(n)):
        if n[j] > 0:
            ar1[possitive_idx] = n[j]
            possitive_idx += 2
        else:
            ar1[negetive_idx] = n[j]
            negetive_idx += 2

    return ar1

print(organizeElements([1,2,3,-1,-2,-3]))
```

Find max consecutive elements in an array

```python
def maxConsecutives(nums):
    max_consicutives = 0
    for i in range(len(nums)):
        
        count = 1
        required = nums[i]+1

        while True:
            isFound = False
            for j in range(len(nums)):
                if nums[j] == required:
                    required += 1
                    count += 1
                    isFound = True
                    break
            if not isFound:
                break
        max_consicutives = max(max_consicutives,count)
    return max_consicutives

print(maxConsecutives([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))
```
