Set all the elements of current row and all the current columns to zero is  matrix element is == 0

```python
def markRow(m,r):
    for i in range(len(m)):
        if i == r:
            for j in range(len(m[i])):
                if m[i][j] != 0:
                    m[i][j] = -1
    return m
def markCol(m,c):
    for i in range(len(m)):
        for j in range(len(m[i])):
            if j == c :
                if m[i][j] != 0:
                    m[i][j] = -1

    return m

def mainMark(m):
    for i in range(len(m)):
        for j in range(len(m[i])):
            if m[i][j] == 0:
                markCol(m,j)
                markRow(m,i)
    for a in range(len(m)):
        for b in range(len(m[a])):
            if m[a][b] == -1:
                m[a][b] = 0
    return m

matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]

print(mainMark(matrix))
```

```python
def matrixx(m):
    colums = [0]*len(m[0])
    rows = [0]*len(m)
    for i in range(len(m)):
        for j in range(len(m[i])):
            if m[i][j] == 0:
                colums[j] = 1
                rows[i] = 1
    for k in range(len(m)):
        for l in range(len(m[k])):
            if colums[l]== 1 or rows[k] ==1:
                m[k][l] = 0

            
    return m

print(matrixx([[1,0]]))
```

```python
def matrixx(matrix):
    
    first_row_zero=False
    first_col_zero=False

    for i in matrix[0]:
        if i == 0:
            first_row_zero= True
            break
    for i in range(len(matrix)):
        if matrix[i][0] == 0:
            first_col_zero = True
            break
    for i in range(1,len(matrix)):
        for j in range(1,len(matrix[i])):
            if matrix[i][j] == 0:
                matrix[0][j] = 0
                matrix[i][0] = 0
  
    for k in range(1,len(matrix)):
        for l in range(1,len(matrix[k])):
            if matrix[k][0]== 0 or matrix[0][l] == 0:
                matrix[k][l] = 0
    if first_col_zero:
        for i in range(len(matrix)):
            matrix[i][0] = 0
    if first_row_zero:
        for i in range(len(matrix[0])):
            matrix[0][i] = 0
    return matrix

print(matrixx([[0,1,2,0],
               [3,4,5,2],
               [1,3,1,5]]))
```
