Given an array of positive integers nums and a positive integer target, return the minimal length of a subarray whose sum is greater than or equal to target. If there is no such subarray, return 0 instead

```python
def generateSubarr(target,nums):

    left = 0
    minlen = float('inf')
    currentSum = 0


    for right in range(len(nums)):
        currentSum += nums[right]


        while currentSum >= target:
            minlen = min(minlen,right - left +1)
            currentSum -= nums[left]
            left += 1
    return 0 if minlen == float('inf') else minlen
    
print(generateSubarr(4,[1,4,4]))
print(generateSubarr(11,[1,1,1,1,1,1,1,1]))
print(generateSubarr(7,[2,3,1,2,4,3]))
print(generateSubarr(11,[1,2,3,4,5]))
     
```

* Merge two sorted arrays wwithout using etra spaces

```python
nums1 = [1,2,3,5,8,0,0,0]
m = 5
nums2 = [2,5,6]
n = 3

def merge( nums1, m, nums2, n):
    mid = m-1
    last2 = n-1
    last  = m + n - 1
    while mid > 0 and last > 0:
         if nums1[mid] < nums2[last2]:
             nums1[last] = nums2[last2]
             last -= 1
             last2 -= 1
         else:
            nums1[last] = nums1[mid]
            mid -= 1
            last -= 1
    return nums1

print(merge(nums1,m,nums2,n))


    
```

Search in a 2d sorted matrix

```python
def searchMatrix(matrix, target):
    for row in matrix:
        l,r = 0,len(row)-1
        if row[l] <= target <= row[r]:
            while l <= r:
                mid = (l+r)//2
                if row[mid] == target:
                    return True
                if row[mid] > target:
                    r = mid -1
                else:
                    l = mid + 1
    return False
print(searchMatrix([[1,3,5,7],[10,11,16,20],[23,30,34,60]],3))
                
```

Count no of inversion in an array

```python
def merge_arr(arr,l,m,h):
    count = 0
    new = []
    left = l
    right = m + 1
    while left <= m and right <= h:
        if arr[left] > arr[right]:
            new.append(arr[right])
            count += (m - left + 1)
            right += 1
        else:
            new.append(arr[left])
            left += 1
    while left <= m:
        new.append(arr[left])
        left += 1
    while right <= h:
        new.append(arr[right])
        right += 1
    for i in range(l,h+1):
        arr[i] = new[i-l]
    return count

def merge_sort(arr,l,h):
    count = 0
    if l >= h:
        return count
    mid = (l + h) // 2

    count += merge_sort(arr,l,mid)
    count += merge_sort(arr,mid+1,h)
    count += merge_arr(arr,l,mid,h)

    return count
ary = [5,4,3,2,1]
print(merge_sort(ary,0,len(ary)-1))
```

```python
def findKthLargest(nums, k):
    tr = len(nums) - k
    def quick_select(nums,l,h,tr):
        if l >= h:
            return nums[l]
        p = quick_sort(nums,l,h)
        if p == tr:
            return nums[p]
        elif p < tr:
            return quick_select(nums,p+1,h,tr)
        elif p > tr:
             return quick_select(nums,l,p-1,tr)
            

    def quick_sort(nums,left,right):
        i = left
        j = right
        p = nums[left]
        while i < j:
            while i <= j and nums[i] <= p:
                i += 1
            while j >= i and nums[j] > p:
                j -= 1
            if i < j:
                nums[i],nums[j] = nums[j],nums[i]
        nums[left],nums[j] = nums[j],nums[left]
        return j
    return quick_select(nums,0,len(nums)-1,tr)
    

print(findKthLargest([3,2,1,5,6,4],2))
            
            


      
```

```python
def permute(nums):
        new = []
        n = len(nums)
        def helper(index,current):
            if index == n:
                if len(current) == n:
                    new.append(current)
                return
            current.append(nums[index])
            helper(current)
            for i in range(index+1,n):
                if nums[i] not in current:
                    current.append(nums[i])
                    helper(i+1,current)
                    # current.pop()
            current.pop()
            

        helper(0,[])
        return new

print(permute([1,3,4,5]))
```
