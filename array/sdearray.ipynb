* Find the next permutation(31)

```python
def generate(nums):
    def reverse(arr,left,right):
        l = left
        r = right
        while l < r:
            nums[l],nums[r] = nums[r],nums[l]
            l += 1
            r -= 1

    n = len(nums)
    pivot = -1
    for i in range(n-2,-1,-1):
        if nums[i] < nums[i+1]:
            pivot = i
            break
    if pivot == -1:
        reverse(nums,0,n-1)
    else:
        minval = float('inf')
        index = -1
        for i in range(n-1,pivot,-1):
            if nums[i] > nums[pivot] and nums[i] <minval:
                index = i
                minval = nums[i]
        nums[pivot],nums[index] = nums[index],nums[pivot]
        reverse(nums,pivot+1,n-1)
    return nums

print(generate([1,3,2]))
            
```

```python
def mergeinter(intervals):
    new = []
    inter = sorted(intervals)

    for i in range(len(inter)):
        if not new or inter[i][0] > new[-1][1]:
            new.append(inter[i])
        else:
            if inter[i][0] <= new[-1][1]:
                new[-1][1] = max(inter[i][1],new[-1][1])
    return new
print(mergeinter([[1,4],[2,3]]))

```

```python

def haveConflict(event1, event2):
    element1 = int(event1[1].replace(":","").strip())
    element2 = int(event2[0].replace(":","").strip())
    if element2 > element1:
        return False
    else:
        return True


print(haveConflict(["10:00","11:00"],["14:00","15:00"]))

      
```
